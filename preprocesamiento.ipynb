Nodos

```python
#Librerias
import pandas as pd
```

```python
#Importing nodes from csv files
df1 = pd.read_csv('data/raw/got-s1-nodes.csv')
df2 = pd.read_csv('data/raw/got-s2-nodes.csv')
df3 = pd.read_csv('data/raw/got-s3-nodes.csv')

print('Season 1 nodes: \n', df1.head(3))
print('\n Season 2 nodes: \n', df1.head(3))
print('\n Season 3 nodes: \n', df1.head(3))
print('\n Sizes \n Season 1: ', df1.shape, '\n Season 2:', df2.shape, '\n Season 3: ', df3.shape,)
```

```python
# Merge Nodes from both seasons
df_nodes = pd.concat([df1,df2,df3], ignore_index=True)

print('Merged Nodes from both seasons: \n', df_nodes.head(3))
print('\n Size of the merged dataframe: ', df_nodes.shape)
```

```python
#Deleting Duplicates that show on both seasons
df_nodes.drop_duplicates(inplace=True)
print('Merged Nodes for both seasons after deleting duplicates: \n', df_nodes.head(3))
print('\n Size of the merged dataframe after deleting duplicates: ', df_nodes.shape)
```

```python
#Saving as a new csv file on the processed folder
df_nodes.to_csv('data/processed/got-nodes.csv', index=False)
```

Relaciones


```python
#Importing edges from csv files
df4 = pd.read_csv('data/raw/got-s1-edges.csv')
df5 = pd.read_csv('data/raw/got-s2-edges.csv')
df6 = pd.read_csv('data/raw/got-s3-edges.csv')


print('Season 1 relationships: \n', df4.head(3))
print('\n Season 2 relationships: \n', df5.head(3))
print('\n Season 3 relationships: \n', df6.head(3))
print('\n Sizes \n Season 1: ', df4.shape, '\n Season 2:', df5.shape, '\n Season 3:', df6.shape)
```

```python
#renaming weight column name to weight_s1
df4.rename(columns={'Weight': 'Weight_s1'}, inplace=True)
df4.drop(columns = ['Season'], inplace = True)

#renaming weight column name to weight_s2
df5.rename(columns={'Weight': 'Weight_s2'}, inplace=True)
df5.drop(columns = ['Season'], inplace = True)

#renaming weight column name to weight_s3
df6.rename(columns={'Weight': 'Weight_s3'}, inplace=True)
df6.drop(columns = ['Season'], inplace = True)


print('Renamed Dataframes \n', df4.head(3), '\n', df5.head(3), '\n', df6.head(3))
```

```python
#Merging the three dataframes into one while keeping all weight columns and filling NaN values with 0
df_edges = pd.merge(df4, df5, on=['Source', 'Target'], how='outer').fillna(0)
df_edges = pd.merge(df_edges, df6, on=['Source', 'Target'], how='outer').fillna(0)
print('Merged Edges Dataframe: \n', df_edges.head(3))
print('\n Size of the merged edges dataframe: ', df_edges.shape)

print('Final Edges Dataframe: \n', df_edges.head(3))
```

```python
#Adding a new column with the total weight of the relationships across all seasons
df_edges['Total_Weight'] = df_edges['Weight_s1'] + df_edges['Weight_s2'] + df_edges['Weight_s3']
print('Edges Dataframe with Total Weight: \n', df_edges.head(3))
```

```python
#Saving as a new csv file on the processed folder
df_edges.to_csv('data/processed/got-edges.csv', index=False)
```
