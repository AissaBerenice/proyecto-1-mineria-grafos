Nodos

```python
#Librerias
import pandas as pd
```

```python
# Importación de los nodos desde el csv
df1 = pd.read_csv('data/raw/got-s1-nodes.csv')
df2 = pd.read_csv('data/raw/got-s2-nodes.csv')

print('Season 1 nodes: \n', df1.head(3))
print('\n Season 2 nodes: \n', df1.head(3))
print('\n Sizes \n Season 1: ', df1.shape, '\n Season 2:', df2.shape)
```

```python
# Merger Nodes from both seasons
df = pd.concat([df1,df2], ignore_index=True)

print('Merged Nodes from both seasons: \n', df.head(3))
print('\n Size of the merged dataframe: ', df.shape)
```

```python
#Deleting Duplicates that show on both seasons
df.drop_duplicates(inplace=True)
print('Merged Nodes for both seasons after deleting duplicates: \n', df.head(3))
print('\n Size of the merged dataframe after deleting duplicates: ', df.shape)
```

```python
#Saving as a new csv file on the processed folder
df.to_csv('data/processed/got-nodes.csv', index=False)
```
